```python
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
import glob
import os
```

```python
input_folder = "C:/python_course/final_project/eunbi/T13/"
output_file = os.path.join(input_folder, "T13_GG_성남시_종합.parquet")

csv_files = glob.glob(os.path.join(input_folder, "T13_GG_PURPOSE_SEXAGE_ADMI_OD_*_성남시.csv"))
csv_files.sort()

chunk_size = 100000
header = ['ETL_YMD', 'O_ADMI_CD', 'O_MEGA_NM', 'O_CTY_NM', 'O_ADMI_NM', 'O_CENTER_X', 'O_CENTER_Y', 'D_ADMI_CD', 'D_MEGA_NM', 'D_CTY_NM', 'D_ADMI_NM', 'D_CENTER_X', 'D_CENTER_Y', 'PURPOSE', 'SEX_CD', 'AGE_GRP', 'CNT'] 
dtype = {
    'ETL_YMD': str,
    'O_ADMI_CD': str,
    'O_MEGA_NM': str,
    'O_CTY_NM': str,
    'O_ADMI_NM': str,
    'O_CENTER_X': 'float64',
    'O_CENTER_Y': 'float64',
    'D_ADMI_CD': str,
    'D_MEGA_NM': str,
    'D_CTY_NM': str,
    'D_ADMI_NM': str,
    'D_CENTER_X': 'float64',
    'D_CENTER_Y': 'float64',
    'PURPOSE': str,
    'SEX_CD': str,
    'AGE_GRP': 'int16',
    'CNT': 'float64'
}

parquet_schema = pa.schema([
    ('ETL_YMD', pa.string()),
    ('O_ADMI_CD', pa.string()),
    ('O_MEGA_NM', pa.string()),
    ('O_CTY_NM', pa.string()),
    ('O_ADMI_NM', pa.string()),
    ('O_CENTER_X', pa.float64()),
    ('O_CENTER_Y', pa.float64()),
    ('D_ADMI_CD', pa.string()),
    ('D_MEGA_NM', pa.string()),
    ('D_CTY_NM', pa.string()),
    ('D_ADMI_NM', pa.string()),
    ('D_CENTER_X', pa.float64()),
    ('D_CENTER_Y', pa.float64()),
    ('PURPOSE', pa.string()),
    ('SEX_CD', pa.string()),
    ('AGE_GRP', pa.int16()),
    ('CNT', pa.float64())
])

parquet_writer = None

for file_path in csv_files:
    print(f"현재 처리 중인 파일: {os.path.basename(file_path)}")

    csv_stream = pd.read_csv(file_path, chunksize=chunk_size, names=header, dtype=dtype, low_memory=False, skiprows=1)

    for chunk in csv_stream:
            # 첫 파일의 첫 청크일 때만 writer 초기화
            if parquet_writer is None:
                parquet_writer = pq.ParquetWriter(output_file, parquet_schema, compression='snappy')
            
            table = pa.Table.from_pandas(chunk, schema=parquet_schema)
            parquet_writer.write_table(table)

if parquet_writer:
    parquet_writer.close()
    print(f"최종 합병 완료! 파일 경로: {output_file}")
```

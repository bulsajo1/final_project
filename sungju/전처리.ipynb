```python
import pandas as pd
import glob
import os
import gzip

base_folder = r'C:\Users\cute3\OneDrive\Desktop\bus_Data\경기도 정류소 별 승하차 인원 집계_2024_D-45'

for month in range(1, 13):  # ⭐ 5월부터 시작
    folder = os.path.join(base_folder, f'CARD_STATION45_2023{month:02d}')
    
    files = glob.glob(os.path.join(folder, '*.csv.gz'))
    
    df_list = []

    for f in files:
        try:
            with gzip.open(f, 'rt', encoding='cp949') as file:
                temp = pd.read_csv(file)
            
            date_str = os.path.basename(f).replace('CARD_STATION45_', '').replace('.csv.gz', '')
            temp['bus_date'] = date_str
            
            df_list.append(temp)

        except Exception as e:
            print(f"에러 파일: {f}")
            print(e)

    # 👉 파일 없는 경우 방지
    if len(df_list) == 0:
        print(f"{month}월 데이터 없음")
        continue

    df = pd.concat(df_list, ignore_index=True)

    output_path = os.path.join(folder, f'bus_2023{month:02d}_all.csv')
    df.to_csv(output_path, index=False, encoding='utf-8-sig')

    print(f"{month}월 완료:", output_path)
```

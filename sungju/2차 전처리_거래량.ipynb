```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
pd.set_option('display.float_format', '{:.0f}'.format)
```

```python
file_path = "../Data/transactions_total.csv"

print(os.path.exists(file_path))

trans_df  = pd.read_csv(file_path, encoding = 'cp949')

trans_df .head(2)
```

```python
trans_df['층'] = trans_df['층'].astype('object')
trans_df.loc[(trans_df['유형'] == '집합') & (trans_df['층'].isna()), '층'] = 'unknown'
trans_df.loc[(trans_df['유형'] == '일반') & (trans_df['층'].isna()), '층'] = 'whole_building'
```

```python
trans_df.loc[(trans_df['유형'] == '집합') & (trans_df['층'] == 'unknown')]
```

```python
# 상·하위 10% 중에서도 극단인 값
lower_10 = np.percentile(trans_df['거래금액(만원)'], 10)
upper_90 = np.percentile(trans_df['거래금액(만원)'], 90)

upper_subset = trans_df[trans_df['거래금액(만원)'] >= upper_90]
q1_u = upper_subset['거래금액(만원)'].quantile(0.25)
q3_u = upper_subset['거래금액(만원)'].quantile(0.75)
iqr_u = q3_u - q1_u
upper_extreme = upper_subset[upper_subset['거래금액(만원)'] > (q3_u + 1.5 * iqr_u)]

lower_subset = trans_df[trans_df['거래금액(만원)'] <= lower_10]
q1_l = lower_subset['거래금액(만원)'].quantile(0.25)
q3_l = lower_subset['거래금액(만원)'].quantile(0.75)
iqr_l = q3_l - q1_l
lower_extreme = lower_subset[lower_subset['거래금액(만원)'] < (q1_l - 0.5 * iqr_l)] 
# q1_l - 1.5 * iqr_l일 경우, 하위 10% 중에서도 극단인 값이 0건으로 나타남. 따라서 하위 극단값 기준을 q1_l - 0.5 * iqr_l로 완화하여 적용함.
```
